```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df=pd.read_csv('https://github.com/YBIFoundation/Live-Projects/raw/main/IndiaHousePrice.csv')
```

```python
df = df.drop_duplicates()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Price'], bins=20, edgecolor='black')
plt.title('Distribution of House Prices')
plt.xlabel('Price')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Living area sqft'], df['Price'], alpha=0.6)

plt.title('Living Area vs House Price')
plt.xlabel('Living Area (sqft)')
plt.ylabel('House Price')

plt.show()
```

```python
print(df.shape)
print(df.columns)
print(df.head())
```

```python
X = df.drop('Price', axis=1)
y = df['Price']
print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=9216
)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test)

print("Predicted house prices:")
print(y_pred[:5])
```

```python
print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
print("India House Price Prediction Project Completed!")
print("Model: Linear Regression")
print("Features used:", list(X.columns))
print("Total predictions tested:", len(y_test))
```

```python
print("Project: India House Price Prediction")
print("Algorithm: Linear Regression")
print("Input Features:", list(X.columns))
print("Number of test samples:", len(y_test))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(y_test, y_pred, alpha=0.5)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    'r--'
)

plt.xlabel('Actual House Prices')
plt.ylabel('Predicted House Prices')
plt.title('Actual vs Predicted House Prices')
plt.show()
```

## Conclusion

The **India House Price Prediction** project was undertaken as part of the Business and Data Analytics internship at YBI Foundation, with the objective of exploring the practical application of data analytics and machine learning in estimating residential property prices. The project was implemented using Python in the Google Colab environment, integrating essential data analysis, visualization, and predictive modelling techniques.

The project commenced with dataset acquisition and exploratory data analysis to examine the structure, attributes, and characteristics of the available data. **Pandas and NumPy** were utilized for data manipulation, numerical operations, and preprocessing. Data quality was assessed by examining duplicate records and checking for missing values. The dataset contained six columns, comprising five predictive features—number of bedrooms, number of bathrooms, living area in square feet, number of nearby schools, and distance from the airport—and one target variable, namely house price. Two duplicate records were identified during the initial assessment and removed during data cleaning.

Subsequently, **Matplotlib and Seaborn** were employed to visualize the distribution of house prices and investigate the relationship between living area and property prices. These visualizations facilitated a clearer understanding of the dataset's characteristics and the potential influence of property-related attributes on price variations.

For predictive modelling, the dataset was divided into training and testing subsets using Scikit-learn's `train_test_split()` function, with 80% of the data allocated for training and 20% reserved for testing. A **Linear Regression** algorithm was then implemented and trained using the selected property features. The trained model generated predictions for the unseen test data through the `predict()` method, enabling an assessment of its generalization performance.

The model's predictive capability was evaluated using three widely recognized regression metrics: Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and the coefficient of determination (R²). The obtained results were as follows:

- **Mean Absolute Error (MAE):** 170,565.47
- **Root Mean Squared Error (RMSE):** 261,345.64
- **R² Score:** 0.5005

The R² score indicates that the model accounts for approximately **50.05% of the variation in house prices within the test dataset**. The MAE and RMSE provide complementary measures of prediction error, with RMSE assigning greater weight to larger errors. These findings indicate that the model captures a meaningful proportion of the observed price variation; however, its predictive performance leaves scope for further improvement. The reported error values are expressed in the same units as the dataset's price variable.

Furthermore, an **Actual vs. Predicted House Prices** scatter plot was generated to visually compare the model's predictions with the corresponding observed values. This visualization provides an intuitive means of assessing prediction discrepancies and identifying instances where predicted prices deviate from actual prices.

In conclusion, the project demonstrates the systematic application of the machine learning workflow, encompassing data acquisition, preprocessing, exploratory analysis, visualization, model training, prediction, and performance evaluation. It also strengthened practical understanding of Python programming, data manipulation, statistical interpretation, and supervised learning within a real-world analytical context.

Future enhancements may include incorporating additional relevant property attributes, evaluating alternative regression algorithms such as Decision Tree Regression and Random Forest Regression, conducting feature engineering, and applying systematic hyperparameter optimization. Such improvements could potentially enhance predictive performance and model reliability.
